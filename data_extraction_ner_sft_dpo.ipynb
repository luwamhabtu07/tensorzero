# Named Entity Recognition with SFT + DPO

This notebook demonstrates how to fine-tune a model for Named Entity Recognition (NER)
using 
**Supervised Fine-Tuning (SFT)** and (later) 
**Direct Preference Optimization (DPO)**.

We'll upload data, fine-tune a model, and evaluate results step by step.

```python
# ✅ Install OpenAI SDK
!pip install --upgrade openai
```

```python
from openai import OpenAI

# ✅ Set up the OpenAI client
client = OpenAI(api_key= "api_key")
```

```python
# ✅ Upload the SFT dataset to OpenAI
with open("data/sft.jsonl", "rb") as f:
    sft_file = client.files.create(file=f, purpose="fine-tune")

print("✅ Uploaded file ID:", sft_file.id)
```

```python
# ✅ Launch the fine-tuning job
sft_job = client.fine_tuning.jobs.create(
    training_file=sft_file.id,

    model="gpt-4o-mini-2024-07-18"  # ✅ This is fine-tune compatible
)

print("🚀 Fine-tune Job ID:", sft_job.id)
```

```python
status = client.fine_tuning.jobs.retrieve(sft_job.id)
print("📊 Job status:", status.status)

if status.status == "succeeded":
    print("✅ Model ID:", status.fine_tuned_model)
```

```python
response = client.chat.completions.create(
    model="ft:gpt-4o-mini-2024-07-18:codeday::C0JVgDM1",
    messages=[
        {"role": "user", "content": "Extract entities from: Elon Musk founded SpaceX in California."}
    ]
)

print("🤖 Model Output:", response.choices[0].message.content)
```
